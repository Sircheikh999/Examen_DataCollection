# Importer packages

```python
# importer packages
import pandas as pd
from selenium.webdriver.common.by import By
import google_colab_selenium as gs
import time
```

# Lancement du navigateur

```python
# Lancer le navigateur
driver = gs.Chrome()
```

# Scraping des données : Source 1 — Books to Scrape


```python
url = 'https://www.gaaraas.com/fr/users/dakar-auto?page=1'
# ouvrir la page
driver.get(url)
```

```python
df_final = pd.DataFrame()

# pages 1 à 13
for i in range(1, 14):

    url = f'https://www.gaaraas.com/fr/users/dakar-auto?page={i}'

    # ouvrir la page
    driver.get(url)
    time.sleep(2)

    # containers = liens vers les annonces
    containers = driver.find_elements(
        By.CSS_SELECTOR,
        'a.common-ad-card'
    )

    # nombre d'annonces sur la page
    number_of_ads = len(containers)

    # récupérer tous les liens avant de changer de page
    ad_urls = []

    for container in containers:
        ad_urls.append(
            container.get_attribute('href')
        )

    data = []

    # parcourir les annonces
    for ad_url in ad_urls:

        try:

            # ouvrir la page de l'annonce
            driver.get(ad_url)

            # récupérer les informations
            dic = {
                'page': i,

                'number_of_ads': number_of_ads,

                'marque_modele': driver.find_element(By.CSS_SELECTOR,'.ad-title-block h2').text,

                'annee': driver.find_element(By.CSS_SELECTOR,'div.prop:nth-of-type(4) span:nth-of-type(2)').text,

                'prix': driver.find_element(By.CSS_SELECTOR,'.back-wrapper .ad-price span.price-wrap').text,

                'kilometrage': driver.find_element(By.CSS_SELECTOR,'div.prop:nth-of-type(3) span:nth-of-type(2)').text,

                'type_boite_de_vitesse': driver.find_element(By.CSS_SELECTOR,'div.prop:nth-of-type(2) span:nth-of-type(2)').text,

                'region_de_vente': driver.find_element(By.CSS_SELECTOR,'.ad-title a span').text,

                'url': ad_url
            }

            data.append(dic)

        except:
            pass

    # dataframe de la page
    df = pd.DataFrame(data)

    # ajouter au dataframe final
    df_final = pd.concat(
        [df_final, df],
        axis=0
    ).reset_index(drop=True)

    print(
        f'Page {i} : {number_of_ads} annonces trouvées | '
        f'{len(df)} annonces récupérées | '
        f'Total : {len(df_final)}'
    )
```
